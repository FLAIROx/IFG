```python
import json
import sys
sys.path.append("/mount/IFG/")
from hendrycks_math.math_eval import compare_final_answer_to_ground_truth, last_boxed_only_string









```

```python
paths = {
    "baseline": "/mount/data_rebuttals/maths_16/baseline/solution_attempts.json",
    "ifg": "/mount/data_rebuttals/maths_16/ifg/solution_attempts.json"
}


data = {}
for key, path in paths.items():
    with open(path, "r") as f:
        data[key] = json.load(f)


```

```python
data['ifg'][0]
```

```python
all_correct = {
    run: sum([all(record['is_correct']) for record in data])
    for run, data in data.items()
}

print(all_correct)
```

```python
any_correct = {
    run: sum([any(record['is_correct']) for record in data])
    for run, data in data.items()
}

print(any_correct)
```

```python
len(data["baseline"]),len(data["ifg"])
```

```python
def all_answers_the_same(record):
    reference = record['attempts'][0]
    for attempt in record['attempts']:
        if not compare_final_answer_to_ground_truth(attempt, reference):
            return False
    return True

```

```python
all_answers_the_same = {
    run: sum(all_answers_the_same(record) for record in data)
    for run, data in data.items()
}

print(all_answers_the_same)
```

```python
first_answers = {
    run: [last_boxed_only_string(record['attempts'][0]) for record in data]
    for run, data in data.items()
}

print(first_answers["baseline"])
print(first_answers["ifg"])
```
