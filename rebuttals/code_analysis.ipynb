```python
import os
import json
import glob
from collections import defaultdict
from statistics import mean

BASE_DIR = "data_rebuttals"

def iter_records_from_file(path):
    if path.endswith(".jsonl"):
        with open(path, "r", encoding="utf-8") as f:
            for line in f:
                line = line.strip()
                if not line:
                    continue
                try:
                    obj = json.loads(line)
                    if isinstance(obj, dict):
                        yield obj
                except json.JSONDecodeError:
                    continue
    else:
        try:
            with open(path, "r", encoding="utf-8") as f:
                obj = json.load(f)
        except json.JSONDecodeError:
            return
        if isinstance(obj, list):
            for item in obj:
                if isinstance(item, dict):
                    yield item
        elif isinstance(obj, dict):
            for key in ("data", "results", "samples", "items"):
                if isinstance(obj.get(key), list):
                    for item in obj[key]:
                        if isinstance(item, dict):
                            yield item
                    return
            for v in obj.values():
                if isinstance(v, list) and v and isinstance(v[0], dict):
                    for item in v:
                        yield item
                    return

def coerce_pass_at_1(value):
    if value is None:
        return None
    if isinstance(value, bool):
        return 1.0 if value else 0.0
    if isinstance(value, (int, float)):
        return float(value)
    if isinstance(value, str):
        try:
            return float(value)
        except ValueError:
            return None
    return None

paths = ['/mount/data_rebuttals/lcb/lcb_with_tuning/baseline_shifted/test/eval_all.json', '/mount/data_rebuttals/lcb/lcb_with_tuning/ifg_shifted/test/eval_all.json']


for p in paths:
    difficulty_to_values = defaultdict(list)
    total_seen = 0
    total_used = 0

    for rec in iter_records_from_file(p):
        total_seen += 1
        difficulty = rec.get("difficulty")
        val = coerce_pass_at_1(rec.get("pass@1"))
        if difficulty is None or val is None:
            continue
        difficulty_to_values[difficulty].append(val)
        total_used += 1

    rel_path = os.path.relpath(p)
    print(f"\nFile: {rel_path}")
    if total_used == 0:
        print("  No valid records with both 'difficulty' and 'pass@1'.")
        print(f"  Records seen: {total_seen}, used: {total_used}, skipped: {total_seen - total_used}")
        continue

    for d in ["easy", "medium", "hard"]:#sorted(difficulty_to_values.keys()):
        vals = difficulty_to_values[d]
        n = len(vals)
        avg_pass = mean(vals)
        avg_gt_001 = sum(1 for v in vals if v > 0.01) / n
        avg_gt_099 = sum(1 for v in vals if v > 0.99) / n
        print(f"  - {d}: pass@1={avg_pass:.4f}, avg[any]={avg_gt_001:.4f}, avg[all]={avg_gt_099:.4f} (n={n})")
    # Print aggregate stats across all difficulties
    all_vals = [v for vals in difficulty_to_values.values() for v in vals]
    n_total = len(all_vals)
    avg_pass_total = mean(all_vals)
    avg_gt_001_total = sum(1 for v in all_vals if v > 0.01) / n_total
    avg_gt_099_total = sum(1 for v in all_vals if v > 0.99) / n_total
    print(f"  - all: pass@1={avg_pass_total:.4f}, avg[any]={avg_gt_001_total:.4f}, avg[all]={avg_gt_099_total:.4f} (n={n_total})")

    print(f"  Records seen: {total_seen}, used: {total_used}, skipped: {total_seen - total_used}")
```

```python
all_error_messages = []
all_error_codes = []
error_messages_all = {}
for run, run_data in data.items():
    error_messages_all[run] = []
    for record in run_data:
        # print([type(i) for i in record['metadata']])
        metadata_decoded =[json.loads(i) for i in record['metadata']]
        error_messages = [x.get('error_message', 'success')  for x in  metadata_decoded]
        all_error_codes.extend([x.get('error_code', 'success')  for x in  metadata_decoded])
        processed_error_messages = []
        for error_message in error_messages:
            if error_message == 'success':
                processed_error_messages.append('success')
            elif "wrong answer" in error_message.lower():
                processed_error_messages.append('wrong_answer')
            elif "time limit" in error_message.lower():
                processed_error_messages.append('timeout')
            elif "error" in error_message.lower():
                processed_error_messages.append('error')
            else:
                raise ValueError(f"Unknown error message: {error_message}")
        error_messages_all[run].append(processed_error_messages)
        all_error_messages.extend(processed_error_messages)

        # assert type(record['metadata']['error_message']) == list       
        # error_messages = [json.loads(x)['error_message']  for x in  eval(record['metadata'])]
        # all_error_messages.extend(error_messages)







pprint(list(set(all_error_messages)))
```

```python
unique_error_messages = list(set(all_error_messages))

for run, run_error_messages in error_messages_all.items():
    print(run)
    for error_message in unique_error_messages:
        n_rows_appears_in = sum(any(i == error_message for i in problem_error_messages) for problem_error_messages in run_error_messages)
        print(f"{error_message} appears in {n_rows_appears_in} rows")
    print("-"*100)
```
